```python
import numpy as np
import matplotlib.pyplot as plt

def equation_holds(sigma_t, t):
    """Check if sigma_t = sigma_t * (1 - t)."""
    if sigma_t == 0:
        return True  # Always holds if sigma_t = 0
    else:
        return np.isclose(t, 0)  # Only holds if t = 0 (for non-zero sigma_t)

# Generate a grid of sigma_t and t values
t_values = np.linspace(-2, 2, 100)
sigma_values = np.linspace(-2, 2, 100)
T, Sigma = np.meshgrid(t_values, sigma_values)

# Evaluate where the equation holds
holds = np.zeros_like(T, dtype=bool)
for i in range(len(sigma_values)):
    for j in range(len(t_values)):
        holds[i, j] = equation_holds(Sigma[i, j], T[i, j])

# Plot the heatmap
plt.figure(figsize=(8, 6))
plt.imshow(holds, extent=[-2, 2, -2, 2], origin='lower', cmap='RdYlGn', alpha=0.7)
plt.colorbar(label='Equation Holds (Green=True, Red=False)')
plt.xlabel('t')
plt.ylabel('σₜ')
plt.title('Validity of σₜ = σₜ (1 - t)')
plt.grid(True, linestyle='--', alpha=0.5)

# Highlight key regions
plt.axhline(0, color='black', linestyle='-', linewidth=1)  # sigma_t = 0 line
plt.axvline(0, color='blue', linestyle='--', linewidth=1, label='t = 0')  # t = 0 line
plt.legend()

plt.show()
```
